# Resultados del Análisis de Mediación

El análisis de mediación se centró en cómo el volumen de trading (variable independiente) influye en el precio promedio (variable dependiente) a través de la volatilidad (variable mediadora). Los resultados del modelo de regresión OLS para la variable mediadora, la volatilidad, indican que la relación entre el volumen de trading y la volatilidad no es significativa. Estos hallazgos revelan que, en este contexto, la variable de volatilidad no actúa como un mediador entre el volumen de trading y el precio promedio. Esta información es valiosa ya que proporciona una perspectiva más clara sobre la dinámica del mercado, destacando que otros factores probablemente influyen en la relación entre el volumen de trading y el precio promedio, más allá de la volatilidad.


# Análisis de Componentes Principales (PCA) para el Mercado de Criptomonedas

Para analizar la compleja interrelación de variables en el mercado de criptomonedas (precio de apertura, cierre, máximo, mínimo, volumen y volatilidad), empleamos el Análisis de Componentes Principales (PCA).  PCA reduce la dimensionalidad de los datos, transformando las variables correlacionadas en un conjunto menor de componentes principales no correlacionadas que explican la mayor parte de la varianza original.  

Visualizamos los datos proyectados en el espacio de los dos primeros componentes principales.  Esto nos permite identificar patrones y tendencias en el mercado, facilitando la interpretación y la identificación de posibles anomalías o eventos inusuales. El porcentaje de varianza explicada por cada componente indica su importancia relativa en la captura de la información contenida en los datos.
